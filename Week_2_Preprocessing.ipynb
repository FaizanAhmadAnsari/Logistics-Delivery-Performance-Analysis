```python
delivery_time_temp = pd.to_datetime(df["delivery_time_hours"])
expected_time_temp = pd.to_datetime(df["expected_time_hours"])

print("Delivery time range:",
      delivery_time_temp.dt.nanosecond.min(),
      "to",
      delivery_time_temp.dt.nanosecond.max())

print("Expected time range:",
      expected_time_temp.dt.nanosecond.min(),
      "to",
      expected_time_temp.dt.nanosecond.max())
```

```python
df["delivery_time_hours_clean"] = delivery_time_temp.dt.nanosecond
df["expected_time_hours_clean"] = expected_time_temp.dt.nanosecond
```

```python
Q1 = df["delivery_time_hours_clean"].quantile(0.25)
Q3 = df["delivery_time_hours_clean"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

```python
outliers = df[df["delivery_time_hours_clean"] > upper_bound]

print("Number of outliers:", len(outliers))
```

```python
print(outliers[[
    "delivery_time_hours_clean",
    "expected_time_hours_clean",
    "distance_km",
    "delivery_mode",
    "delivery_status",
    "delayed"
]].head(20))
```

```python
categorical_columns = [
    "delivery_partner",
    "package_type",
    "vehicle_type",
    "delivery_mode",
    "region",
    "weather_condition",
    "delayed",
    "delivery_status"
]

for col in categorical_columns:
    print("\n", col)
    print(df[col].unique())
```

```python
numerical_columns = [
    "distance_km",
    "package_weight_kg",
    "delivery_rating",
    "delivery_cost"
]

for col in numerical_columns:
    print("\n", col)
    print("Min:", df[col].min())
    print("Max:", df[col].max())
    print("Mean:", df[col].mean())
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

columns_to_normalize = [
    "distance_km",
    "package_weight_kg",
    "delivery_cost"
]

df[[col + "_normalized" for col in columns_to_normalize]] = scaler.fit_transform(
    df[columns_to_normalize]
)
```

```python
print(df[[
    "distance_km",
    "distance_km_normalized",
    "package_weight_kg",
    "package_weight_kg_normalized",
    "delivery_cost",
    "delivery_cost_normalized"
]].head(10))
```
